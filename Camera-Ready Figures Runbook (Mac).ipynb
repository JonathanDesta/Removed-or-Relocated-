Addresses N03, N04, N07, N08, N09, N12 and figure legibility in N14 from the YKSA-11 review.
The notebook calls the same repository scripts as the terminal workflow. It reads saved
records, never loads a model, and creates a fresh versioned output directory. Existing
figures, raw results and checkpoint files remain unchanged.

Select the **colab-local** kernel. Install it once if needed:
```

```
Use **Run All** from a fresh kernel. Any failure is logged and stops the workflow. Restart

the repeated Llama baseline (tau approximately .105) and names that choice explicitly.


````python
import datetime, os, subprocess, sys, uuid
from pathlib import Path
from IPython.display import Image, display

REPO = Path(os.environ.get("PAPER_REPO", "/Users/jonathandesta/Documents/GitHub/Removed or Relocated?"))
PROJECT = Path(os.environ.get("PAPER_PROJECT", str(Path.home() / "Library/CloudStorage/GoogleDrive-jdjonathandesta@gmail.com/My Drive/maheep-yksa")))
PYTHON = Path.home() / ".venvs/colab-local/bin/python"
if Path(sys.prefix).resolve() != (Path.home() / ".venvs/colab-local").resolve():
    raise RuntimeError("Select the colab-local kernel and restart.")
# PAPER_REVISION_OUT is optional; it must name a directory that does not yet exist.
OUTPUT = Path(os.environ.get("PAPER_REVISION_OUT", str(PROJECT / "revisions" /
    ("yksa11-" + datetime.datetime.now(datetime.timezone.utc).strftime("%Y%m%dT%H%M%SZ") + "-" + uuid.uuid4().hex[:8]))))

def run(stage):
    subprocess.run([str(PYTHON), "-u", str(REPO / "scripts/paper_revision.py"), stage,
                    "--project", str(PROJECT), "--out", str(OUTPUT)], cwd=REPO, check=True,
                   env=dict(os.environ, PYTHONDONTWRITEBYTECODE="1", MPLBACKEND="Agg"))

def show(basename):
    display(Image(filename=str(OUTPUT / "figures" / (basename + ".png")), width=768))

run("init")
print("OUTPUT:", OUTPUT)
````
